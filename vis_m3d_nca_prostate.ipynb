We use the selected scans from MRI Prostate dataset from the Medical Segmentation Decathlon. CT scans from the dataset along with a pretrained M3d-NCA model are included with this repository.

### Intensity Mapping Options<br>
In general, the visualization algorithm in the NCA-VIS tool works by segmenting the intensity spectrum of NCA-Channel activations into a number of bins, which are mapped onto segments of equal width in the visualized color spectrum.<br>
The NCA-VIS tool offers a broad range of options to generate intensity mappings. All are situated in this window: <br>
<div>
<img src="src/visualization/assets/all_options.png" width="800"><br><br>
</div>

* **1**: Allows you to choose how intensity values within an individual bin are mapped to colors. The default option is a linear mapping, but you can also choose a tangential mapping. For the tangential mapping, q_values can be chosen via the q_value slider<br><br>
* **2**: Selects the NCA-Channel which is visualized.<br><br>
* **3**: Chooses the granularity of the intensity mapping: If set to "single", a single intensity mapping is generated for all steps of the NCA. If set to "time stepped", an individual intensity mapping is generated for each step of the NCA. <br><br>
* **4**: Controls the number of bins used for the intensity mapping. In general, more bins correspond to more visible detail in the visualization. <br><br>
* **5**: Controls how the bins are generated. "EQUI WIDTH" means, that bins are generated so that each bin is inhabbited by an equal number of values. "MULTI OTSU" is an adaptive binning algorithm that selects bin-borders to minimize intra-bin variance. <br><br>
* **6**: The default setting "ABSOLUTE" is used to display the absolute channel activations. The "DIFF" setting gives the option to display the change in intensity between steps. <br><br>




### Rendering & Visualization

The upper part of the application controls the playback & rendering of the generated visualizations. <br>

<div>
<img src="src/visualization/assets/display_render_options.png" width="800"><br><br>
</div>

* **1**: Allows you to step the visualization along the Z-axis, if a 3D model is used for visualization.<br><br>
* **2**: Allows you to stop playback of the visualization and manually step through the inference steps.<br><br>
* **3**: Option to toggle on/off the intensity mapping. If the intensity mapping is disabled, NCA-activation values are simply cropped to the range [0, 1].<br><br>
* **4**: Renders the currently displayed interpolation step. Opens a file-explorer dialog to select location / filename. Supports rendering as .jpeg and .PNG.<br><br>
* **5**: This button renders the entire inference process as a .GIF.<br><br>
* **6**: This dropdown menu provides different color maps for visualization.<br><br>
* **7**: Options for stepped rendering. Renders a step-by-step version of the inference process as individual .PNGs. The **Render Step** field gives control over the step width (i.e. only every n-th step is rendered).
